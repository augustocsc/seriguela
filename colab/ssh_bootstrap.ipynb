# 🔑 SSH Bootstrap — Debug Remoto do Colab

Cria um túnel SSH via Cloudflare (sem cadastro) para acesso remoto ao ambiente Colab.

**Uso:**
1. Execute a célula abaixo
2. Copie o endpoint exibido (`tcp://xxx.trycloudflare.com:PORTA`)
3. Envie para o Claude no chat
4. Claude se conecta via SSH e pode inspecionar/modificar o ambiente

**Segurança:** A senha é temporária e gerada aleatoriamente. Mude `SSH_PASSWORD` se quiser uma senha específica.

```python
# ═══════════════════════════════════════════════════════════════
# SSH Bootstrap via Cloudflare Tunnel (sem cadastro)
# ═══════════════════════════════════════════════════════════════

import random
import string

# Gera senha aleatória de 8 caracteres
SSH_PASSWORD = ''.join(random.choices(string.ascii_letters + string.digits, k=8))

print(f'Senha SSH: {SSH_PASSWORD}')
print('Instalando colab-ssh...')

!pip install -q colab-ssh

from colab_ssh import launch_ssh_cloudflared

print('\nIniciando túnel SSH...')
print('(Aguarde o endpoint aparecer abaixo)')
print('='*60)

launch_ssh_cloudflared(password=SSH_PASSWORD)

print('='*60)
print(f'\n⚠️  Copie o endpoint acima e envie para o Claude junto com a senha: {SSH_PASSWORD}')
print('O Claude vai se conectar e poderá:')
print('  - Ler logs em tempo real')
print('  - Checar o processo do daemon')
print('  - Editar arquivos e reiniciar processos')
```
